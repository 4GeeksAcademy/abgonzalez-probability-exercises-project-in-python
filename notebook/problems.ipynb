## Exercise 1 

Two dices are thrown once and the total score is observed. Use a simulation to find the estimated probability that the total score is even or greater than 7.  A simulation is a repetition of the same experiment multiple times to observe its behavior:

- Run the experiment 1000 times (roll 2 dice 1000 times, and sum the number of both dices).
- Keep track of the number of times that the sum was either greater than 7 or an even number.
- Divide the number from step 2 by the number of iterations (1000).

```python
import numpy as np

n_simulations = 1000

count = 0

for i in range(n_simulations):
    dice1 = np.random.randint(1, 7)
    dice2 = np.random.randint(1, 7)
    
    total = dice1 + dice2
    
    if total % 2 == 0 or total > 7:
        count += 1

# Calculate probability
probability = count / n_simulations

print(f"Number of successful outcomes: {count}")
print(f"Estimated probability: {probability}")
print(f"Estimated probability (percentage): {probability * 100:.2f}%")
```

## Exercise 2

A box contains 10 white balls, 20 red balls and 30 green balls. If we take 5 balls from the box with replacement (we take the ball, observe what color it is and put it back into the box). We want to know the probability of:

1. Take 3 white and 2 red.
2. Take all of the same color.

Run the experiment 1000 times and calculate the above probabilities.

```python
import numpy as np

ball_box = {}

# Create the box of balls
for i in range(60):
    if i < 10:
        ball_box[i] = "White"
    elif (i > 9) and (i < 30):
        ball_box[i] = "Red"
    else:
        ball_box[i] = "Green"

print(ball_box)


n_simulations = 1000

count_3white_2red = 0
count_all_same_color = 0

for i in range(n_simulations):
    draws = []
    for j in range(5):
        # We always check with 60 balls, even we pick up one.
        ball_index = np.random.randint(0, 60)
        draws.append(ball_box[ball_index])
    
    white_count = draws.count("White")
    red_count = draws.count("Red")
    green_count = draws.count("Green")
    
    if white_count == 3 and red_count == 2:
        count_3white_2red += 1
    
    if white_count == 5 or red_count == 5 or green_count == 5:
        count_all_same_color += 1

prob_3white_2red = count_3white_2red / n_simulations
prob_all_same = count_all_same_color / n_simulations

print(f"Probability of 3 white and 2 red: {prob_3white_2red:.4f} ({prob_3white_2red * 100:.2f}%)")
print(f"Number of times: {count_3white_2red} out of {n_simulations}")
print(f"Probability of all same color: {prob_all_same:.4f} ({prob_all_same * 100:.2f}%)")
print(f"Number of times: {count_all_same_color} out of {n_simulations}")
```

### 1. Qué es combinatoria, permutación y como esto se relaciona con probabilidad.

Combinacion:
- Es un subconjunto de elementos de un conjunto dado, donde el orden no importa.
- Se utiliza para contar el número de formas en que se pueden seleccionar r elementos de un conjunto
Probabilidad de un evento -> P(evento)=  casos favorables/casos totales
Probabilidad de elegir un k elementos de un total de n elementos cuando el orden no improta
C(n,k) = n! / (k!(n-k)!) 

​Permutacion:
- Es un arreglo de elementos de un conjunto dado, donde el orden si importa.
P(n)=n!

1.1. Diferencia entre en combinatoria o permutación.
Combinatoria: El orden no importa, se utiliza para contar el número de formas en que se pueden seleccionar r elementos de un conjunto.
Permutación: El orden si importa, se utiliza para contar el número de formas en que se pueden arreglar n elementos.
